# Chapter 04: Introduction to Neural Learning — Gradient Descent

**Primary Reference:**
  
Andrew W. Trask, 
*Grokking Deep Learning*
Manning Publications

## 1. Chapter Summary

**Neural learning** enables neural networks to refine their predictions through error evaluation and systematic weight adjustments driven by feedback from training data.

This chapter introduces **gradient descent** as a learning method that determines the direction and magnitude of weight updates for reducing prediction errors.

Key concepts such as derivatives, learning rates, and divergence provide the foundation for optimizing neural networks through systematic error minimization.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Evaluating the accuracy of neural network predictions.
- Measuring prediction errors during training.
- Applying hot-and-cold learning.
- Calculating weight update direction and magnitude.
- Understanding gradient descent optimization.
- Reducing errors through iterative learning.
- Applying derivatives in neural learning.
- Understanding divergence and the impact of learning rates.

# Chapter 15: Deep Learning on Unseen Data — Introducing Federated Learning

**Primary Reference:**
  
Andrew W. Trask, 
*Grokking Deep Learning*
Manning Publications

## 1. Chapter Summary

**Federated learning** enables deep learning models to be trained across distributed datasets without requiring sensitive information to be centralized, transferred, or directly exposed to external parties.

This chapter examines privacy challenges in deep learning, federated training workflows, secure aggregation strategies, and homomorphic encryption techniques that support collaborative model development.

Experiments with secure aggregation and encrypted model updates demonstrate how federated learning can preserve privacy while maintaining effective model training.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Examining privacy challenges in deep learning.
- Understanding federated learning principles.
- Applying neural networks to spam detection.
- Analyzing vulnerabilities in federated learning systems.
- Understanding secure aggregation techniques.
- Exploring homomorphic encryption concepts.
- Applying homomorphic encryption to federated learning.

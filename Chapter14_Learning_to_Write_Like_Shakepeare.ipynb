# Chapter 14: Learning to Write like Shakepeare — Long Short-Term Memory

**Primary Reference:**
  
Andrew W. Trask, 
*Grokking Deep Learning*
Manning Publications

## 1. Chapter Summary

**Long short-term memory (LSTM) networks** enhance recurrent neural networks by incorporating memory and gating mechanisms that improve the retention of relevant information across extended sequences.

This chapter examines character-level language modeling, truncated backpropagation, gradient instability, and recurrent learning dynamics, highlighting limitations encountered when training conventional recurrent networks.

LSTM cells reduce vanishing and exploding gradients through gated memory mechanisms, enabling more robust sequence learning and the generation of coherent language patterns.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Applying character-level language modeling.
- Understanding truncated backpropagation techniques.
- Analyzing vanishing and exploding gradients.
- Examining backpropagation in recurrent networks.
- Understanding LSTM cell architectures.

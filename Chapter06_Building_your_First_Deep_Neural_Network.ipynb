# Chapter 06: Building your First Deep Neural Network — Introduction to Backpropagation

**Primary Reference:**
  
Andrew W. Trask, 
*Grokking Deep Learning*
Manning Publications

## 1. Chapter Summary

**Deep neural networks** introduce hidden layers that enable models to learn complex relationships beyond the direct correlations captured by single-layer architectures.

This chapter explains **backpropagation** as a method that propagates prediction errors through network layers, enabling weights to be adjusted according to their contribution to model outputs.

Through matrix computations, nonlinear transformations, and gradient-based optimization, illustrate how deep neural networks learn representations that capture increasingly complex data patterns.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Applying the streetlight problem to neural learning.
- Understanding matrices and matrix relationships.
- Comparing full, batch, and stochastic gradient descent.
- Understanding how neural networks learn correlation.
- Understanding overfitting its effects.
- Creating correlations through hidden layers.
- Applying backpropagation for error attribution.
- Distinguishing linear and nonlinear networks.
- Understanding conditional correlation in deep learning.
- Building deep neural network architectures.
- Implementing backpropagation in code.

### 3.2 What are Parametric Models and Nonparametric Models?

#### 3.2.1 Concept Explanation

**Parametric and nonparametric learning** represent two machine learning approaches that differ in how information is stored and how model complexity is determined.

**Parametric models** rely on a fixed number of adjustable parameters that are refined during training, whereas **nonparametric models** allow complexity to increase according to data.

These approaches influence how patterns are represented, with parametric methods emphasizing parameter optimization and nonparametric methods relying more heavily on observed examples.

#### Mathematical Representation

A **parametric model** generates predictions using a function with a fixed number of parameters that are adjusted during the learning process:

$$
\hat{y} = f(x;\theta)
$$

| Symbol | Description |
| :--- | :--- |
| $\hat{y}$ | Predicted output |
| $x$ | Input data |
| $f$ | Learned function |
| $\theta$ | Learned model parameters with a fixed dimensionality |

A **nonparametric model** generates predictions using information derived from observed data, where the number of parameters can change according to the dataset:

$$
\hat{y} = f_D(x)
$$

| Symbol | Description |
| :--- | :--- |
| $\hat{y}$ | Predicted output |
| $x$ | Input data |
| $D$ | Observed training dataset |
| $f_D$ | Learned function determined by the available training data |

The nonparametric model retrieves results directly from stored observations, causing its complexity and information storage requirements to depend on available data.

This comparison highlights the fundamental distinction discussed in the chapter, where parametric methods emphasize **parameter adjustment while nonparametric methods depend more heavily on data representation.**

#### 3.3.1 Concept Explanation

**Supervised and unsupervised learning** represent two major categories of machine learning that differ in **how input data is transformed into outputs.**

**Supervised learning** learns patterns between input and output datasets to make predictions, whereas **unsupervised learning** identifies patterns and groups within a dataset without predefined target outputs.

#### Mathematical Representation

A **supervised learning** learns a relationship between input and target datasets, allowing it to predict the corresponding output based on the observed patterns:

$$
\hat{y} = f_\theta(x)
$$

| Symbol | Description |
| :--- | :--- |
| $\hat{y}$ | Predicted output |
| $x$ | Input data |
| $f_\theta$ | Learned mapping function parameterized by $\theta$ |
| $\theta$ | Learned model parameters |

An **unsupervised learning** transforms input data into an output that represents discovered patterns or groups without relying on predefined target outputs:

$$
z = g_\theta(x)
$$

| Symbol | Description |
| :--- | :--- |
| $z$ | Discovered pattern/learned representation |
| $x$ | Input data |
| $g_\theta$ | Learned pattern discovery function |
| $\theta$ | Learned model parameters |

### 3.4 How Can Machines Learn?

#### 3.4.1 Concept Explanation

**Machine learning** can be viewed as a process of prediction, evaluation, and improvement that enables systems to learn patterns from experience.

This learning process commonly consists of prediction, comparison, and adjustment stages that gradually refine the model’s ability to capture patterns within data.

Through these stages, machine learning models adjust their parameters based on prediction errors to produce more accurate outputs for similar inputs.

#### Mathematical Representation

A model generates a **prediction** by combining input data with its current parameters before comparing the resulting output with the expected outcome:

$$
\hat{y} = f(x;\theta)
$$

| Symbol | Description |
| :--- | :--- |
| $\hat{y}$ | Predicted output |
| $x$ | Input data |
| $f$ | Model function |
| $\theta$ | Currrent model parameters |

In the **error comparison** stage, the predicted output is evaluated against the actual outcome to determine how closely the model reproduces the target pattern:

$$
Error = y - \hat{y}
$$

| Symbol | Description |
| :--- | :--- |
| $y$ | Actual output/target value |
| $\hat{y}$ | Predicted output |
| $Error$ | Difference between actual and predicted outputs |

During the **learning update**, model parameters are adjusted using the prediction error and input data to improve future predictions on similar observations:

$$
\theta_{\text{new}} = Update(\theta_{\text{old}},x,Error) 
$$

| Symbol | Description |
| :--- | :--- |
| $\theta_{\text{new}}$ | Updated model parameters |
| $\theta_{\text{old}}$ | Previous model parameters |
| $x$ | Input data used during prediction |
| $Error$ | Difference between actual and predicted outputs |
| $Update$ | Parameter adjustment process based on the input data and prediction error |

#### 3.4.2 Code Reproduction

This section does not contain direct code reproduction because the original material explains the **Predict–Compare–Learn process** conceptually without providing a direct executable code implementation.

#### Additional Demonstration

The following demonstration illustrates how a model can adjust its parameter based on the prediction error and input data.

```python
input_value = 2
parameter = 0.40 #knob 
actual_value = 1.00

prediction = input_value * parameter
error = actual_value - prediction

learning_rate = 0.10 #parameter adjustment speed
parameter = parameter + (learning_rate * error * input_value)

updated_prediction = input_value * parameter

data = {
    "Initial Prediction": prediction,
    "Prediction Error": error,
    "Updated Parameter": parameter,
    "Updated Prediction": updated_prediction
}

width = max(len(label) for label in data)

for label, value in data.items():
    print(f"{label:<{width}} : {value:.3f}")
```

#### 3.4.3 Code Explanation

The code initializes an `input_value` and a model **`parameter`** to generate an initial **`prediction`** using a simple multiplication operation.

The **`prediction`** is compared with the `actual_value` and the resulting **`error`** is used alongside the input and learning rate to adjust the **`parameter`**.

The **`updated parameter`** is then used to calculate a new **`prediction`**, illustrating how changing model parameters can influence subsequent outputs.

#### 3.4.4 Result Analysis & Interpretation

The initial **`prediction`** is 0.80, while the `actual_value` is 1.00, producing an **`error`** of 0.20 that indicates the model's prediction is below the target.

After the **`parameter`** is adjusted from 0.40 to 0.44, the **`updated prediction`** increases to 0.88, moving closer to the actual value.

This behavior reflects the **Predict–Compare–Learn cycle** described in the chapter, where models gradually improve through repeated adjustments.

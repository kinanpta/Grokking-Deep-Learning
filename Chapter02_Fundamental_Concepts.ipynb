# Chapter 02: Fundamental Concepts — How Do Machines Learn?

**Primary Reference:**
  
Andrew W. Trask, 
*Grokking Deep Learning*
Manning Publications

## 1. Chapter Summary

**Machine learning models** uncover patterns by adjusting parameters or structures based from examples, enabling predictions and insights without relying on explicitly programmed rules.

This chapter differentiates parametric and nonparametric models, supervised and unsupervised learning, and clarifies the roles of inputs, targets, and model assumptions.

The **predict‑compare‑learn cycle** is presented as a structured process that generates predictions, measures errors, and improves model performance through iterative feedback and parameter updates.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Differentiating deep learning, machine learning, and artificial intelligence.
- Comparing parametric and nonparametric models by characteristics and complexity.
- Contrasting supervised and unsupervised learning applications.
- Explaining how machines learn through data, predictions, and model adjustments.

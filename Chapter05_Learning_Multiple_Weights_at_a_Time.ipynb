## 1. Chapter Summary

**Neural networks with multiple weights** integrate information from several inputs, enabling models to learn more sophisticated relationships than those achievable with single-weight prediction systems.

This chapter **extends gradient descent** to architectures with multiple inputs and outputs, explaining how individual weight updates contribute to overall learning performance.

Weight deltas, dot products, and coordinated parameter updates illustrate how multiple weights interact to improve prediction accuracy across increasingly complex neural network architectures.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Applying gradient descent to networks with multiple inputs.
- Analyzing the effects of freezing individual weights.
- Applying gradient descent to networks with multiple outputs.
- Training multi-input and multi-output networks.
- Visualizing weight changes throughout the learning process.
- Interpreting dot products in neural networks.

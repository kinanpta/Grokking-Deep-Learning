## 1. Chapter Summary

**Convolutional Neural Networks (CNN)** reuse the same weights across multiple input locations, reducing parameter counts while improving the ability to learn generalizable visual patterns.

This chapter introduces convolutional layers as structures that detect recurring features, such as edges and corners, by applying shared kernels throughout images.

Experiments with image classification demonstrate how weight sharing improves feature detection, reduces overfitting, and increases predictive performance compared with fully connected architectures.

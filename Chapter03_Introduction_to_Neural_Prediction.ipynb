# Chapter 03: Introduction to Neural Prediction — Forward Propagation

**Primary Reference:**
  
Andrew W. Trask, 
*Grokking Deep Learning*
Manning Publications

## 1. Chapter Summary

**Forward propagation** demonstrates how neural networks transform input values into predictions by combining adjustable weights and passing signals through interconnected layers of computation.

This chapter starts with basic prediction problems, progressively developing intuition for single‑weight and multi‑input computations using clear numerical examples to illustrate the mechanics.

Each computational step highlights how parameters affect outputs, providing insight into network behavior before learning algorithms adjust weights during subsequent training phases.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Implementing simple neural networks for prediction tasks.
- Explaining neural network structure and prediction process.
- Calculating predictions using multiple inputs and weights.
- Understanding neural networks producing multiple outputs.
- Combining inputs and outputs in single network architecture.
- Explaining layer predictions as inputs for subsequent layers.

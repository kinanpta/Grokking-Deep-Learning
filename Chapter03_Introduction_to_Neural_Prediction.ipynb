# Chapter 03: Introduction to Neural Prediction — Forward Propagation

**Primary Reference:**
  
Andrew W. Trask, 
*Grokking Deep Learning*
Manning Publications

## 1. Chapter Summary

**Forward propagation** demonstrates how neural networks transform input values into predictions by combining adjustable weights and passing information through interconnected layers of computation.

This chapter begins with basic prediction tasks progressively developing intuition for single‑weight and multi‑input computations using clear numerical examples to illustrate the mechanics.

Each computational step highlights how parameters influence predictions, providing insight into network behavior before learning algorithms adjust weights during training.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Implementing simple neural networks for prediction tasks.
- Explaining neural network structure and prediction process.
- Calculating predictions using multiple inputs and weights.
- Understanding neural networks with multiple outputs.
- Combining multiple inputs and outputs within a single architecture.
- Explaining how layer outputs become inputs for subsequent layers.

## 1. Chapter Summary

**Activation functions** allow neural networks to capture nonlinear patterns, providing the flexibility needed to learn complex relationships that cannot be represented through linear operations alone.

This chapter examines hidden-layer and output-layer activation functions, including sigmoid, tanh, and softmax, while explaining their roles in prediction and probability modeling.

Activation derivatives, output interpretations, and implementation techniques demonstrate how appropriate activation functions improve learning behavior and classification performance in neural networks.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Understanding activation functions in neural networks.
- Comparing sigmoid and tanh activation functions.
- Applying softmax for probability prediction.
- Implementing activation functions in neural networks.

## 4. Discussion & Conclusion

This chapter demonstrated how **activation functions introduce non-linearities** into neural networks, enabling models to form complex boundaries and output formatted probability distributions.

Utilizing **sigmoid and tanh activation functions**, alongside **softmax output layers**, transforms raw network outputs into interpretable probabilities suitable for multi-class classification tasks.

The successful technical implementation of activation layers establishes versatile classification networks, providing essential preparation for spatial feature learning in Convolutional Neural Networks in Chapter 10.

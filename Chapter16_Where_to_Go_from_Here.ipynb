# Chapter 16: Where to Go from Here — A Brief Guide

**Primary Reference:**
  
Andrew W. Trask, 
*Grokking Deep Learning*
Manning Publications

## 1. Chapter Summary

This concluding chapter outlines **practical next steps for deep learning study, emphasizing continued experimentation, broader technical foundations, and participation in research communities.** 

It recommends learning established frameworks such as PyTorch, strengthening mathematics, implementing research papers, and sharing knowledge through writing and collaboration. 

These activities help learners move from reproducing textbook examples toward independent projects, professional development, and sustained engagement with the field.

## 2. Learning Objectives

This chapter aims to provide an understanding of:

- Identifying appropriate next steps after learning fundamental deep learning concepts.
- Recognizing the value of established frameworks such as PyTorch.
- Understanding the importance of strengthening mathematical foundations.
- Exploring additional courses, textbooks, and advanced learning resources.
- Developing skills through reproducing and implementing academic research.
- Recognizing the role of computational resources, including GPUs.
- Understanding the benefits of technical writing, open-source contributions, and community participation.
- Planning continued learning through independent projects and practical experimentation.
